```python
import pandas as pd
import matplotlib.pyplot as plt

# Load IMDb dataset into a DataFrame
url = "https://raw.githubusercontent.com/PacktPublishing/Pandas-Cookbook-Second-Edition/master/data/movie_metadata.csv"
df = pd.read_csv(url)

# Display first 5 rows
print(df.head())

plt.figure(figsize=(6,4))
plt.plot(df["Movie"], df["Collection"], marker='o')
plt.title("Movie Collection")
plt.xlabel("Movie")
plt.ylabel("Collection (Crores)")
plt.grid(True)
plt.show()


plt.figure(figsize=(6,4))
plt.bar(df["Movie"], df["Rating"])
plt.title("Movie Ratings")
plt.xlabel("Movie")
plt.ylabel("Rating")
plt.show()

plt.figure(figsize=(6,6))
plt.pie(df["Collection"],
        labels=df["Movie"],
        autopct="%1.1f%%",
        startangle=90)
plt.title("Collection Share")
plt.show()

plt.figure(figsize=(6,4))
plt.hist(df["Rating"], bins=5)
plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df["Rating"], df["Collection"])
plt.title("Rating vs Collection")
plt.xlabel("Rating")
plt.ylabel("Collection (Crores)")
plt.show()

plt.figure(figsize=(4,5))
plt.boxplot(df["Collection"])
plt.title("Collection Box Plot")
plt.ylabel("Collection (Crores)")
plt.show()
```
