```python
# ============================================================
# EXPERIMENT 4
# DATA INSPECTION, FILTERING, STATISTICAL ANALYSIS, CORRELATION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("DataCoSupplyChainDataset.csv", encoding="latin1")

print("Dataset Loaded Successfully")

# ============================================================
# PROGRAM 1 - DATA INSPECTION
# ============================================================

print("\nFirst 5 Records")
print(df.head())

print("\nLast 5 Records")
print(df.tail())

# ============================================================
# PROGRAM 2 - SHAPE
# ============================================================

print("\nShape of Dataset")
print(df.shape)

print("\nRows =", df.shape[0])
print("Columns =", df.shape[1])

# ============================================================
# PROGRAM 3 - DATA TYPES
# ============================================================

print("\nData Types")
print(df.dtypes)

print("\nDataset Info")
print(df.info())

# ============================================================
# PROGRAM 4 - FILTERING
# Shipping Cost > 500
# ============================================================

high_shipping = df[df["shipping_costs"] > 500]

print("\nShipping Cost > 500")
print(high_shipping.head())

# ============================================================
# PROGRAM 5 - FILTER HIGH RISK
# ============================================================

high_risk = df[df["risk_classification"] == "High Risk"]

print("\nHigh Risk Records")
print(high_risk.head())

# ============================================================
# PROGRAM 6 - SORTING
# ============================================================

sorted_df = df.sort_values(
    by="shipping_costs",
    ascending=False
)

print("\nTop Shipping Costs")
print(sorted_df.head(10))

# Bar Chart

top10 = sorted_df.head(10)

plt.figure(figsize=(10,5))

plt.bar(range(len(top10)), top10["shipping_costs"])

plt.xticks(range(len(top10)), top10["supplier_id"], rotation=90)

plt.xlabel("Supplier")

plt.ylabel("Shipping Cost")

plt.title("Top 10 Shipping Costs")

plt.show()

# ============================================================
# PROGRAM 7 - STATISTICAL ANALYSIS
# ============================================================

print("\nMean =", df["shipping_costs"].mean())

print("Median =", df["shipping_costs"].median())

print("Mode =")
print(df["shipping_costs"].mode())

print("Standard Deviation =", df["shipping_costs"].std())

print("Minimum =", df["shipping_costs"].min())

print("Maximum =", df["shipping_costs"].max())

# ============================================================
# PROGRAM 8 - DESCRIPTIVE STATISTICS
# ============================================================

print(df.describe())

# ============================================================
# PROGRAM 9 - GROUPING
# ============================================================

group = df.groupby("risk_classification")["shipping_costs"].mean()

print("\nAverage Shipping Cost by Risk")

print(group)

# Visualization

group.plot(kind="bar", figsize=(7,5))

plt.xlabel("Risk Classification")

plt.ylabel("Average Shipping Cost")

plt.title("Average Shipping Cost by Risk")

plt.show()

# ============================================================
# PROGRAM 10 - AGGREGATION
# ============================================================

summary = df.groupby("risk_classification").agg({

    "shipping_costs":"mean",

    "lead_time_days":"mean",

    "historical_demand":"mean"

})

print(summary)

# ============================================================
# PROGRAM 11 - CORRELATION
# ============================================================

numeric = df.select_dtypes(include=np.number)

corr = numeric.corr()

print(corr)

plt.figure(figsize=(12,8))

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()

# ============================================================
# PROGRAM 12 - SCATTER PLOT
# ============================================================

plt.figure(figsize=(7,5))

plt.scatter(df["lead_time_days"],
            df["shipping_costs"])

plt.xlabel("Lead Time Days")

plt.ylabel("Shipping Costs")

plt.title("Lead Time vs Shipping Cost")

plt.show()

# ============================================================
# PROGRAM 13 - HISTOGRAM
# ============================================================

plt.figure(figsize=(7,5))

plt.hist(df["shipping_costs"], bins=30)

plt.xlabel("Shipping Cost")

plt.ylabel("Frequency")

plt.title("Shipping Cost Distribution")

plt.show()

# ============================================================
# PROGRAM 14 - BOXPLOT
# ============================================================

plt.figure(figsize=(7,5))

sns.boxplot(x=df["shipping_costs"])

plt.title("Shipping Cost Boxplot")

plt.show()

# ============================================================
# PROGRAM 15 - VALUE COUNTS
# ============================================================

print(df["risk_classification"].value_counts())

# ============================================================
# PROGRAM 16 - UNIQUE VALUES
# ============================================================

print(df["supplier_country"].unique())

# ============================================================
# FINAL REPORT
# ============================================================

print("\n==============")

print("Experiment Completed Successfully")

print("Dataset Shape :", df.shape)

print("Missing Values :", df.isnull().sum().sum())

print("Correlation Matrix Generated")

print("==============")
```
