```python
import pandas as pd
import os
import kagglehub

# Download latest version
path = kagglehub.dataset_download("START-UMD/gtd")

print("Path to dataset files:", path)
# Load the terrorist dataset already downloaded using kagglehub
csv_file = [f for f in os.listdir(path) if f.endswith(".csv")][0]

df = pd.read_csv(
    os.path.join(path, csv_file),
    encoding="latin1",
    low_memory=False
)

# Display selected terrorist dataset columns
data = {
    "Year": df["iyear"].head(5).tolist(),
    "Country": df["country_txt"].head(5).tolist(),
    "Region": df["region_txt"].head(5).tolist(),
    "Attack_Type": df["attacktype1_txt"].head(5).tolist()
}

terrorist_df = pd.DataFrame(data)

print("Original DataFrame:")
print(terrorist_df)

print("\n1. First 5 rows (head):")
print(terrorist_df.head())

print("\n2. Last 2 rows (tail):")
print(terrorist_df.tail(2))

print("\n3. Shape (rows, columns):")
print(terrorist_df.shape)

print("\n4. Column Names:")
print(terrorist_df.columns)

print("\n5. DataFrame Information:")
terrorist_df.info()

print("\n6. Statistical Summary:")
print(terrorist_df.describe())

print("\n7. Attacks after 2010:")
print(
    df[df["iyear"] > 2010][
        ["iyear", "country_txt", "region_txt", "attacktype1_txt"]
    ].head()
)
```
