```python
import pandas as pd
data = {
    "Name": ["Ashi", "Rahul", "Priya", "Kiran", "Neha"],
    "Age": [20, 21, 19, 22, 20],
    "City": ["Chennai", "Bangalore", "Hyderabad", "Delhi", "Mumbai"]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
print("\n1. First 5 rows (head):")
print(df.head())
print("\n2. Last 2 rows (tail):")
print(df.tail(2))
print("\n3. Shape (rows, columns):")
print(df.shape)
print("\n4. Column Names:")
print(df.columns)
print("\n5. DataFrame Information:")
df.info()
print("\n6. Statistical Summary:")
print(df.describe())
print("\nAge > 20:")
print(df[df["Age"] > 20])
```
